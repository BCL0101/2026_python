```python
#高中數學
'''
Homework(degree.py)
讓使用者輸入直角三角形的對邊
讓使用者輸入直角三角形的斜邊
計算角度
公式:
sin(x) = 對邊 / 斜邊
x是radian(弧度)
'''

import math

opposite = float(input("請輸入直角三角形的對邊："))
hypotenuse = float(input("請輸入直角三角形的斜邊："))

if opposite > hypotenuse:
    print("錯誤：對邊不能大於斜邊，請重新輸入。")
else:
    sin_x = opposite / hypotenuse
    x_radian = math.asin(sin_x)

    # 使用 round() 進行四捨五入，並將度數轉換為整數
    print("角度（弧度）=", round(x_radian, 2))
    print("角度（度數）=", int(round(math.degrees(x_radian), 0))) #如果傳出值不要有小數點就要加入int()
```
