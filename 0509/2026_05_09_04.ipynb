```python
#學生總分為300
#有些學生可以加分5%
scores:int|float = int(input("請輸入學生分數(最高300分)")) #:int, typehint, 型別提示, |<<<加了這個的意思是"或"（沒有強制性）
is_add:str = input("學生是否符合加分條件？(y,n)")

if is_add == "y":
  scores = scores * 1.05 #程式區塊不能是空的,如果暫時沒填東西,可先填pass
  if scores > 300: #if裡面還有if=巢狀判斷
    scores = 300

print("學生分數是：", round(scores, 2))
```
