# 數學運算

```python
# 建立變數

10 #(int)
10.0 #浮點數(float)
"10.5" #字串(str)
'10.5' #字串(str)
```

```python
print(10 + 5)

print(10.0 + 5.0)

print("10.0 + 5.0")

print(10.0 + 5)
```

```python
10 + 5 #運算式(expression)
```

```python
a = 5
b = 10
print(a)
print(b)
print(a + b)
```

```python
a = 10
b = 3

print(a + b)   # 13, 加法
print(a - b)   # 7, 減法
print(a * b)   # 30, 乘法
print(a / b)   # 3.3333333333333335, 除法
print(a // b)  # 3, 整除
print(a % b)   # 1, 取餘數
print(a ** b)  # 1000, 次方
```

```python
a = 5
b = 10
total = a + b
print(total)
print(99 ** 99)
```

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 * (1 + 2) ** 2 - 2 ** 2 * 2)
```
