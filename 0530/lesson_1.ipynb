```python
# 資料結構
# 程式結構

# list
# 可以修改
scores = [10, 20, 30, 40, 50]
print(scores)
print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
print(scores[4])
scores[0] = 100
print(scores)
```

```python
# tuple
# tuple建立後, 不可以修改內容, read only
# 何時使用, 暫時儲存
scores1 = (10, 20, 30, 40, 50)
print(scores1)
print(scores1[0])
print(scores1[1])
print(scores1[2])
print(scores1[3])
print(scores1[4])
scores1[0] = 100
print(scores1)
```
