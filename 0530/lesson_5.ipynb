```python
file = open("student.txt", "r", encoding="utf-8")
content = file.read()
file.close()  # 👈 必須手動寫這行，把門關上
```

```python
with open ("student.txt","r",encoding="utf-8") as file:
    content = file.read()
# 這裡發生任何事情，不論是順利結束還是突然噴錯（Error）...
    
# 👈 只要程式的「縮排」一結束、離開了 with 的勢力範圍
# Python 就會自動在後台幫你執行 file.close()！   
print(file.closed)
```
