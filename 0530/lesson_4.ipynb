```python
# open("students.txt") 引數值的呼叫

try:
	with open("student.txt", "r", encoding="utf-8") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
# open(file="students.txt") 引數名稱的呼叫

try:
	with open(file="student.txt") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
# open("students.txt") 引數值的呼叫, 必須依順序

try:
	with open("student.txt", 'r', -1, 'utf-8') as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
# open(file="student.txt") 引數名稱的呼叫,可以不依順序,只要依照名稱

try:
	with open(file="student.txt",encoding="utf-8") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
# open(file="student.txt") 混合的呼叫,可以不依順序,只要依照名稱

try:
	with open("student.txt",encoding="utf-8",mode="r") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
class WineBottle:  # 類別：酒瓶設計圖
    def __init__(self, name, target_volume):
        # 這裡定義的就是「屬性」
        self.name = name                 # 酒的名字
        self.volume = target_volume      # 當前剩餘毫升數

    # 這裡定義的就是「實體方法」
    def pour(self, ml):  # 倒酒的動作
        if self.volume >= ml:
            self.volume -= ml  # 動作會去改變自己的「屬性」！
            print(f"從【{self.name}】倒出了 {ml}ml，還剩下 {self.volume}ml。")
        else:
            print(f"【{self.name}】不夠倒了！只剩 {self.volume}ml。")

# --- 進入酒館實作 ---

# 1. 煉製一個實體
vodka = WineBottle("伏特加", 700)

# 2. 讀取它的「屬性」（名詞，沒括號）
print(vodka.name)    # 輸出：伏特加
print(vodka.volume)  # 輸出：700

# 3. 呼叫它的「實體方法」（動詞，有括號）
vodka.pour(100)      # 輸出：從【伏特加】倒出了 100ml，還剩下 600ml。
```
