```python
# dictionary
students = {'Chinese':10, 'English':20, 'Math':30, 'History':40, 'Discover':50}
print(students)
print(students['Chinese'])
print(students['Math'])
print(students['Discover'])
print(students['English'])
print(students['History'])
```

```python
# list
scores1 = [10, 20, 30, 40, 50]
for s in scores1:
  print(s)
```

```python
# dict
students = {'Chinese':10, 
            'English':20, 
            'Math':30, 
            'History':40, 
            'Discover':50}

for key in students:   # 使用for in來提取特定資料
    print(students[key])
```
