```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['PingFang TC']
plt.rcParams['axes.unicode_minus'] = False

labels = ['奶茶', '紅茶', '綠茶', '咖啡']
colors = ['yellow', 'red', 'green', 'brown']

values = []
for drink in labels:
    num = int(input(f'請輸入{drink}的銷售杯數：'))
    values.append(num)

max_value = max(values)
max_index = values.index(max_value)

explode = [0.2 if i == max_index else 0 for i in range(len(values))]

figure = plt.figure(figsize=(7, 7))
axes = figure.add_subplot()

axes.pie(values,
         labels=labels,
         colors=colors,
         explode=explode,
         startangle=90,
         shadow=True,
         autopct='%1.1f%%',
         labeldistance=1.1,
         pctdistance=0.7,
         textprops={'color': 'black'})

axes.axis('equal')
plt.show()
```
