```python
a = [1, 2, 3]
print(a + [4, 5])   # [1, 2, 3, 4, 5]
```

```python
import numpy as np

b = np.array([1, 2, 3])   # array 就是「專門拿來做數值運算的多維資料結構」 ndarray才能做陣列運算
print(b + np.array([4, 5, 6]))   #[5 7 9]
```
