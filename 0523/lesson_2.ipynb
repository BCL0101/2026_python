```python
for i in range(0, 5, 2):   # range(start, stop, step=1), *次數計算不包含stop本身, 例如 5 就只會從 0 算到 4 
    print(f"第 {i+0} 次")   # f 是字串插補 f-string 
```

```python
score = int(input("請輸入分數:"))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("不及格!!")
```

```python
score = int(input("請輸入分數:"))

# 建立分數與等級的對照表 (由高到低排列)
grades = {90: "A", 80: "B", 70: "C", 60: "D"}

# 預設為不及格，只要滿足對照表門檻就更新並中斷
result = "不及格!!"
for threshold, grade in grades.items():
    if score >= threshold:
        result = grade
        break

print(result)
```
