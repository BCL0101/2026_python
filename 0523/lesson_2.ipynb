```python
for i in range(0, 5, 2):   # range(start, stop, step=1), *次數計算不包含stop本身, 例如 5 就只會從 0 算到 4 
    print(f"第 {i+0} 次")   # f 是字串插補 f-string 
```

```python
# 數字加總的冗長方法
i = 0
i += 1
i += 2
i += 3
i += 4
i += 5
i += 6
i += 7
i += 8
i += 9
i += 10
print(i)
```

```python
# 直接用 range 作為加總的簡化
total = 0

for i in range(1,11):
  total += i

print(total)
```
