```python
# 學生總分為300
# 有些學生可以加分5%

while True:   #無限迴圈
    try:  # try except指令是為了預防出錯
        scores = int(input("請輸入學生分數(最高300分):"))
        break   #跳出迴圈
    except ValueError:
        print("輸入錯誤，請輸入整數！")

is_add = input("學生是否符合加分條件?(y,n)")

if is_add == "y": #單項選擇
	scores = scores * 1.05

# 四捨五入(5進位)
scores = int(scores + 0.5)

# 超過300分，上限取300
if scores > 300:
	scores = 300

print("學生分數:",scores)
```

```python
i = 1

while i <= 10:   #while後面一定要有":"<這個叫做布林值, 後面跟著的是程式區塊
    print(i)
    i += 2   #這個+=叫做複合指定運算子
```

```python
i = 1

while i <= 10:
    i += 1
    if i % 2 == 0:   #把變數 i 拿去除以 2，看得到的餘數是不是剛好等於 0, 是的話回到 while 繼續執行迴圈
        continue
    print(i)   #如果餘數不 = 0, 將會傳出 i 值
```

```python
i = 0
while i < 10:
    i += 1

    if i == 3: 
        continue   #如果 i 沒有 = 3,就會再回傳到 while 繼續執行

    if i == 8:   #當 i = 8 時, 觸發 break 結束迴圈
        break

    print(i)   
    
```
