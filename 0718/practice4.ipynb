```python
import numpy as np
from numpy import ndarray
import pandas as pd
from pandas import DataFrame

scores:ndarray | None = None # 學生的分數
scores_df:DataFrame | None = None # 學生分數的DataFrame

scores = np.random.randint(50, 101, size=(50, 5))
scores_df = pd.DataFrame(scores,            
            columns=['國文', '英文', '數學', '地理', '歷史'],
            index=range(1,51))

names_df:DataFrame = pd.read_csv('students.csv')
scores_df[['姓名','性別']] = names_df[['姓名','性別']].head(n=50).values
scores_df.groupby("性別")[['國文', '英文', '數學', '地理', '歷史']].agg('mean')
```

```python
scores_df
```

```python
# 依性別分組，計算各科平均分數
subjects = ['國文', '英文', '數學', '地理', '歷史']
gender_group = scores_df.groupby('性別')[subjects].mean().round(2)
gender_group
```

```python
# 計算學生總分，並分成三個分數階級
scores_df['學生總分'] = scores_df[subjects].sum(axis=1)
scores_df['分數階級'] = pd.cut(
    scores_df['學生總分'],
    bins=[0, 300, 400, float('inf')],
    labels=['低分區', '一般區', '高分區'],
    right=False
)
scores_df
```

```python
# 依分數階級分組，統計學生人數與平均總分
score_level_group = (
    scores_df.groupby('分數階級', observed=False)
    .agg(學生人數=('姓名', 'count'), 平均總分=('學生總分', 'mean'))
    .round(2)
    .sort_index(ascending=False)
)
score_level_group
```
